## Q22

````python
a = "hello"
b = "".join(["he", "llo"])
print(a == b)
print(a is b)
````


## Q35

````python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)
````

**Answer**

```text
Output:
[[1, 99], [3, 4]]
```

**Explanation:** A shallow copy duplicates only the outer list. The inner lists are shared.

## Q36

````python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)
````

**Answer**

```text
Output:
[[1, 2], [3, 4]]
```

**Explanation:** A deep copy recursively copies the inner lists too, so changing b does not affect a.
